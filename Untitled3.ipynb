```python
from datetime import datetime

class digitclock:
    def _init_(self):
        pass
    def showtime(self):
        now = datetime.now()
        print ("เวลาปัจจุบัน : {}".format(now.strftime("%H:%M:%S")))
```

```python
clock = digitclock()
clock.showtime()
```

```python
class parent:
    def _init_(self):
        pass

    def hello(self):
        print("หวัดดี")
```

```python
class child(parent):
    def _init_(self):
        pass
```

```python
person = child()
person.hello()
```
